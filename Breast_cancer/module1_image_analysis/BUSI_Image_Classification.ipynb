# BUSI Breast Ultrasound Image Classification

This notebook uses ImageNet-pretrained MobileNetV2 for three-class classification (Normal, Benign, Malignant), with a frozen feature extractor and a trainable classification head. The optional training cell is commented out; no training occurs unless you explicitly uncomment it.

The dataset contains 780 original images (133 Normal, 437 Benign, 210 Malignant). Segmentation masks are excluded. Splits are stratified at the image level; no patient IDs are used, so these are not patient-level splits.

## 1. Connect Google Drive and configure the dataset path
Place `Dataset_BUSI_with_GT` in Drive. Its expected structure is one folder per class, each containing original images and optional mask files.

```python
from google.colab import drive
drive.mount('/content/drive')

from pathlib import Path

# Update this path if your BUSI dataset is stored elsewhere in Drive.
DATASET_DIR = Path('/content/drive/MyDrive/Dataset_BUSI_with_GT')
CLASS_NAMES = ['normal', 'benign', 'malignant']
IMAGE_SIZE = (224, 224)
BATCH_SIZE = 32
SEED = 42

if not DATASET_DIR.is_dir():
    raise FileNotFoundError(f'BUSI dataset directory not found: {DATASET_DIR}')
print('Dataset:', DATASET_DIR)
```

## 2. Find original images and exclude masks
Files whose names contain `_mask` are segmentation masks and are not included. The count assertions help catch an incorrect Drive path or unexpected dataset contents before any training.

```python
import numpy as np
from sklearn.model_selection import train_test_split

IMAGE_EXTENSIONS = {'.png', '.jpg', '.jpeg', '.bmp'}
image_paths, labels = [], []
for label, class_name in enumerate(CLASS_NAMES):
    class_dir = DATASET_DIR / class_name
    if not class_dir.is_dir():
        raise FileNotFoundError(f'Missing class directory: {class_dir}')
    originals = sorted(
        p for p in class_dir.iterdir()
        if p.is_file() and p.suffix.lower() in IMAGE_EXTENSIONS
        and '_mask' not in p.stem.lower()
    )
    image_paths.extend(originals)
    labels.extend([label] * len(originals))

image_paths = np.asarray([str(p) for p in image_paths])
labels = np.asarray(labels, dtype=np.int32)
counts = {name: int(np.sum(labels == i)) for i, name in enumerate(CLASS_NAMES)}
print('Original image counts:', counts)
print('Total original images:', len(image_paths))
expected = {'normal': 133, 'benign': 437, 'malignant': 210}
if counts != expected or len(image_paths) != 780:
    raise ValueError(f'Expected 780 originals with counts {expected}; found {len(image_paths)} and {counts}. Check DATASET_DIR.')
```

## 3. Stratified train, validation and test splits
The split is 70% train, 15% validation and 15% test, stratified by class. It is image-level because patient identifiers are not available in this dataset layout. If multiple images come from the same patient, this split does not prevent that patient appearing in more than one subset.

```python
train_paths, temp_paths, y_train, y_temp = train_test_split(
    image_paths, labels, test_size=0.30, random_state=SEED, stratify=labels
)
val_paths, test_paths, y_val, y_test = train_test_split(
    temp_paths, y_temp, test_size=0.50, random_state=SEED, stratify=y_temp
)

def show_split_counts(name, y):
    print(f'{name}: {len(y)} images | ' + ', '.join(
        f'{CLASS_NAMES[i]}={int(np.sum(y == i))}' for i in range(len(CLASS_NAMES))
    ))
show_split_counts('Train', y_train)
show_split_counts('Validation', y_val)
show_split_counts('Test', y_test)
```

## 4. Decode, resize, preprocess and batch images
Images are decoded as RGB, resized to 224×224, then transformed with MobileNetV2's `preprocess_input` (pixel values mapped to the range expected by ImageNet-pretrained weights). The input pipeline reads from Drive on demand and batches samples for Colab.

```python
import tensorflow as tf

AUTOTUNE = tf.data.AUTOTUNE
NUM_CLASSES = len(CLASS_NAMES)

def decode_resize_preprocess(path, label):
    image_bytes = tf.io.read_file(path)
    image = tf.io.decode_image(image_bytes, channels=3, expand_animations=False)
    image.set_shape([None, None, 3])
    image = tf.image.resize(image, IMAGE_SIZE, antialias=True)
    image = tf.keras.applications.mobilenet_v2.preprocess_input(image)
    return image, tf.one_hot(label, NUM_CLASSES)

def make_dataset(paths, y, training=False):
    ds = tf.data.Dataset.from_tensor_slices((paths, y))
    if training:
        ds = ds.shuffle(buffer_size=len(paths), seed=SEED, reshuffle_each_iteration=True)
    ds = ds.map(decode_resize_preprocess, num_parallel_calls=AUTOTUNE)
    return ds.batch(BATCH_SIZE).prefetch(AUTOTUNE)

train_ds = make_dataset(train_paths, y_train, training=True)
val_ds = make_dataset(val_paths, y_val)
test_ds = make_dataset(test_paths, y_test)
print('Batched datasets are ready.')
```

## 5. Basic training-only augmentation
Random horizontal/vertical flips and small rotations are applied only while the model is training. Augmentation layers are inactive during validation and inference. Ultrasound orientation assumptions should be reviewed before interpreting results.

```python
data_augmentation = tf.keras.Sequential([
    tf.keras.layers.RandomFlip(mode='horizontal_and_vertical'),
    tf.keras.layers.RandomRotation(factor=0.04),
], name='data_augmentation')
```

## 6. MobileNetV2 transfer-learning classifier
The ImageNet-pretrained MobileNetV2 base uses `include_top=False` and stays frozen initially. A global average pooling layer, dropout and three-class softmax head are trained first. MobileNetV2 receives images preprocessed in the input pipeline above.

```python
from sklearn.utils.class_weight import compute_class_weight

# Balanced weights are computed from the training split only.
class_ids = np.arange(NUM_CLASSES)
weight_values = compute_class_weight(class_weight='balanced', classes=class_ids, y=y_train)
CLASS_WEIGHTS = {int(class_id): float(weight) for class_id, weight in zip(class_ids, weight_values)}
print('Class weights:', {CLASS_NAMES[k]: v for k, v in CLASS_WEIGHTS.items()})

def build_mobilenetv2_classifier(input_shape=(224, 224, 3), num_classes=3):
    base = tf.keras.applications.MobileNetV2(
        input_shape=input_shape, include_top=False, weights='imagenet'
    )
    base.trainable = False

    inputs = tf.keras.Input(shape=input_shape, name='ultrasound_image')
    x = data_augmentation(inputs)
    x = base(x, training=False)
    x = tf.keras.layers.GlobalAveragePooling2D(name='global_average_pooling')(x)
    x = tf.keras.layers.Dropout(0.30, name='head_dropout')(x)
    outputs = tf.keras.layers.Dense(num_classes, activation='softmax', name='class_probabilities')(x)
    model = tf.keras.Model(inputs, outputs, name='busi_mobilenetv2_classifier')
    model.compile(
        optimizer=tf.keras.optimizers.Adam(learning_rate=1e-3),
        loss='categorical_crossentropy',
        metrics=['accuracy']
    )
    return model, base

model, mobilenet_base = build_mobilenetv2_classifier()
print('MobileNetV2 base trainable:', mobilenet_base.trainable)
model.summary()
```

## 8. Classification report and confusion matrix
After training, this cell reports test loss and accuracy, plus class-wise precision/recall/F1 and macro/weighted averages, and plots a confusion matrix. It expects a trained model; the test set should remain held back until final evaluation.

```python
from sklearn.metrics import classification_report, confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt

def evaluate_model(trained_model):
    print('Test metrics:', trained_model.evaluate(test_ds, return_dict=True, verbose=0))
    probabilities = trained_model.predict(test_ds, verbose=0)
    y_true = np.concatenate([np.argmax(batch_y.numpy(), axis=1) for _, batch_y in test_ds])
    y_pred = probabilities.argmax(axis=1)
    print(classification_report(y_true, y_pred, target_names=[n.title() for n in CLASS_NAMES], digits=4, zero_division=0))
    cm = confusion_matrix(y_true, y_pred, labels=np.arange(NUM_CLASSES))
    display = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=[n.title() for n in CLASS_NAMES])
    display.plot(cmap='Blues', values_format='d', xticks_rotation=30)
    plt.title('BUSI Test Confusion Matrix')
    plt.tight_layout()
    plt.show()
    return probabilities, y_true, y_pred

# After training only: test_probabilities, y_true, y_pred = evaluate_model(model)
```

## 9. Confidence-based uncertain predictions
The helper returns `Uncertain` whenever the highest class probability is below a configurable threshold. Set the threshold using validation data and a documented operating criterion; the example value 0.60 is only a starting point, not a clinically validated cutoff.

```python
UNCERTAIN_THRESHOLD = 0.60  # Tune using validation data; not clinically validated.

def predict_with_uncertainty(trained_model, image_batch, threshold=UNCERTAIN_THRESHOLD):
    probs = trained_model.predict(image_batch, verbose=0)
    results = []
    for row in probs:
        top_index = int(np.argmax(row))
        confidence = float(row[top_index])
        label = CLASS_NAMES[top_index].title() if confidence >= threshold else 'Uncertain'
        results.append({'prediction': label, 'confidence': confidence, 'class_probabilities': dict(zip(CLASS_NAMES, map(float, row)))})
    return results

# Example after training: predict_with_uncertainty(model, one_image_batch, threshold=0.60)
```
